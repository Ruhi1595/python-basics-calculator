# Simple Python Calculator
Building a calculator includes 3 main steps:
1. **Input:** Accepting any number (whole or decimal) from the user.
2. **Operation:** Choosing the mathematical operator (`+`, `-`, `*`, `/`).
3. **Result:** Calculating and displaying the final output.

```python
# Step 1: Assigning variables and getting input from the user
num1 = float(input("Enter the first number: "))
operator = input("Enter an operator(+, -, /, *): ")
num2 = float(input("Enter the second number: "))
# Step 2: Perform calculation on the chosen operator
if operator == "+":
    result = num1 + num2
elif operator == "-":
    result = num1 - num2
elif operator == "*":
    result = num1 * num2
elif operator == "/":
    if num2 == 0:
        result = "Error! Cannot divide by zero"
    else:
        result = num1 / num2
else:
    result = "Invalid Operator!"

#Display the result
print("Result", result)
```
